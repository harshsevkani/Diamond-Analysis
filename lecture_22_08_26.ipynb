```python
plt.figure(figsize=[12,6])
plt.hist(df1["price"],bins = 200, color="purple")
plt.xlabel("Price in USD")
plt.ylabel("Frequency")
plt.title("Diamonds Price Histogram")
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df1['price'],df1['carat'])
plt.title('Carat v/s Price')
plt.xlabel('Price')
plt.ylabel('Carat')
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df1['price'],df1['depth'])
plt.title('Depth v/s Price')
plt.xlabel('Price')
plt.ylabel('Depth')
plt.show()
```

```python
plt.figure(figsize=(12,6))
cut_count.plot(kind='bar',color='orange')
plt.title('Distribution of Diamond Cuts')
plt.xlabel('Cut Type')
plt.ylabel('Frequency')
plt.xticks(rotation = 45)
plt.grid(axis='y',linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(12,6))
cut_count.plot(kind='pie',autopct='%1.1f%%',color=['skyblue','lightgreen','orange','lightcoral','gold'])
plt.title('Distribution of Diamonds Cuts')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.scatter(df1['x'],df1['y'])
plt.title("X-axis v/s Y-axis")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
